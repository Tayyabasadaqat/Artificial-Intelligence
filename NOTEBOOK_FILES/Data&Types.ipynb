#### Mean:
Sum of all/Count of all
- No chance of outlier (i.e" rare case {in last row salary=508K})
#### Median:
Sort all value then find the middle one
- where chance of outlier exists
#### Mode:
Next frequent value

```python
import statistics as st
def mean(data):
    print("Mean =", st.mean(data))
def median(data):
    print("Median =", st.median(data))
def mode(data):
    print("Mode =", st.mode(data))
```
