```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
'Name':['Ali','Ahmed','Umair','Ayesha','Rahim','Fatima','Hassan'],
'Age':[24,25,21,19,24,22,18],
'Gender':['M','M','M','F','M','F','M'],
'Math':[68,72,98,74,54,57,81],
'Phy':[77,68,96,82,49,38,74],
'Comp':[89,67,89,81,62,54,79]
}
df=pd.DataFrame(data)
df
```

```python
df["Total"] = df["Math"] + df["Phy"] + df["Comp"]
avg_math = df["Math"].mean()
avg_phy = df["Phy"].mean()
avg_comp = df["Comp"].mean()
gender_count = df["Gender"].value_counts()
fig = plt.figure(figsize=(16,9),facecolor='skyblue')
fig.suptitle("Student Performance Dashboard", fontsize=20, fontweight="bold")
plt.subplot(2,3,1)
plt.pie(gender_count,
        labels=gender_count.index,
        autopct='%1.1f%%',
        explode=(0.05,0),
        startangle=90)
plt.title("Gender Distribution")
plt.subplot(2,3,1)
plt.pie(gender_count,
        labels=gender_count.index,
        autopct='%1.1f%%',
        explode=(0.05,0),
        startangle=90)
plt.title("Gender Distribution")
plt.subplot(2,3,3)
subjects = ["Math","Physics","Computer"]
averages = [avg_math, avg_phy, avg_comp]

plt.bar(subjects, averages,
        color=["steelblue","orange","teal"])
plt.title("Average Marks")
plt.ylim(0,100)
plt.subplot(2,3,4)

plt.plot(df["Name"], df["Math"],
         marker='o', linewidth=2, label="Math")

plt.plot(df["Name"], df["Phy"],
         marker='o', linewidth=2, label="Physics")

plt.plot(df["Name"], df["Comp"],
         marker='o', linewidth=2, label="Computer")

plt.title("Marks Comparison")
plt.ylim(0,100)
plt.legend()
plt.subplot(2,3,5)

bars = plt.bar(df["Name"], df["Total"],
               color="gold", edgecolor="black")

plt.title("Total Marks")

for bar in bars:
    h = bar.get_height()
    plt.text(bar.get_x()+bar.get_width()/2,
             h+2,
             str(int(h)),
             ha='center')
plt.subplot(2,3,6)
plt.axis("off")

summary = f"""
DATASET SUMMARY

Students : {len(df)}
Male     : {(df['Gender']=='M').sum()}
Female   : {(df['Gender']=='F').sum()}

Average Age : {df['Age'].mean():.1f}

Average Marks
Math      : {avg_math:.1f}
Physics   : {avg_phy:.1f}
Computer  : {avg_comp:.1f}

Highest Total :
{df.loc[df['Total'].idxmax(),'Name']}
({df['Total'].max()})

Lowest Total :
{df.loc[df['Total'].idxmin(),'Name']}
({df['Total'].min()})
"""

plt.text(0,1,summary,
         fontsize=11,
         va="top",
         family="monospace",
         bbox=dict(facecolor="lavender",
                   edgecolor="black"))

plt.tight_layout()
plt.savefig("plots/dashboard.png", dpi=300, bbox_inches="tight")
plt.show()
```
