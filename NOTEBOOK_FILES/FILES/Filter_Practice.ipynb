```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
import os

print("Current folder:", os.getcwd())
print("Files in folder:")
print(os.listdir())
```

```python
image = cv2.imread("My_image1.jpg")

# Convert BGR to RGB for displaying with Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

print("Image Shape:", image.shape)
```

```python
# 1. Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. Gaussian Blur
gaussian_blur = cv2.GaussianBlur(image, (7, 7), 0)

# 3. Median Blur
median_blur = cv2.medianBlur(image, 7)

# 4. Bilateral Filter
bilateral = cv2.bilateralFilter(image, 9, 75, 75)

# 5. Average Blur
average_blur = cv2.blur(image, (7, 7))

# 6. Sharpening
sharpen_kernel = np.array([
    [0, -1,  0],
    [-1, 5, -1],
    [0, -1,  0]
])

sharpened = cv2.filter2D(image, -1, sharpen_kernel)

# 7. Canny Edge Detection
canny = cv2.Canny(gray, 100, 200)
```

```python
plt.figure(figsize=(20, 10))

# Original
plt.subplot(2, 4, 1)
plt.imshow(image_rgb)
plt.title("Original", fontsize=14, fontweight="bold")
plt.axis("off")

# Grayscale
plt.subplot(2, 4, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale", fontsize=14, fontweight="bold")
plt.axis("off")

# Gaussian Blur
plt.subplot(2, 4, 3)
plt.imshow(cv2.cvtColor(gaussian_blur, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blur", fontsize=14, fontweight="bold")
plt.axis("off")

# Median Blur
plt.subplot(2, 4, 4)
plt.imshow(cv2.cvtColor(median_blur, cv2.COLOR_BGR2RGB))
plt.title("Median Blur", fontsize=14, fontweight="bold")
plt.axis("off")

# Bilateral Filter
plt.subplot(2, 4, 5)
plt.imshow(cv2.cvtColor(bilateral, cv2.COLOR_BGR2RGB))
plt.title("Bilateral Filter", fontsize=14, fontweight="bold")
plt.axis("off")

# Average Blur
plt.subplot(2, 4, 6)
plt.imshow(cv2.cvtColor(average_blur, cv2.COLOR_BGR2RGB))
plt.title("Average Blur", fontsize=14, fontweight="bold")
plt.axis("off")

# Sharpening
plt.subplot(2, 4, 7)
plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
plt.title("Sharpening", fontsize=14, fontweight="bold")
plt.axis("off")

# Canny Edge
plt.subplot(2, 4, 8)
plt.imshow(canny, cmap="gray")
plt.title("Canny Edge", fontsize=14, fontweight="bold")
plt.axis("off")

plt.tight_layout()
plt.savefig("Filters.png")
plt.show()
```
