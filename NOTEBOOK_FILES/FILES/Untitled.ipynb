## Imports

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
print(df.columns)
```

### Line plot

```python
sns.set_style("darkgrid")
sns.lineplot(data=df, x="total_bill", y="tip")
plt.title("Line plot of bill vs tips")
plt.show()
```

#### Bar plot

```python
sns.set_style("darkgrid")
sns.barplot(data=df, x="day", y="total_bill")
plt.title("Bar plot of day vs total bill")
plt.show()
```

#### Count plot

```python
sns.set_style("darkgrid")
sns.countplot(data=df, x="day")
plt.title("Count plot")
plt.show()
```

#### Box Plot 

```python
sns.set_style("darkgrid")
sns.boxplot(data=df, x="day", y="total_bill")
plt.title("Box plot")
plt.show()
```

#### Violin Plot

```python
sns.set_style("darkgrid")
sns.violinplot(data=df, x="day",y="total_bill")
plt.title("Violin plot")
plt.show()
```

####  Strip Plot

```python
sns.set_style("darkgrid")
sns.stripplot(data=df, x="day", y="total_bill")
plt.title("Strip plot")
plt.show()
```

#### Swarm plot

```python
sns.set_style("darkgrid")
sns.swarmplot(data=df, x="day", y="total_bill")
plt.title("Swarm plot")
plt.show()
```

#### Histogram plot

```python
sns.set_style("darkgrid")
sns.histplot(data=df, x="total_bill", bins=50)
plt.title("Histogram Plot")
plt.show()
```

### KDE Plot

```python
sns.set_style("darkgrid")
sns.kdeplot(df["total_bill"], fill=True)
plt.title("KDE Plot")
plt.show()
```

### HeatMap Plot

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(8,6))
sns.heatmap(corr)
plt.title("Correlation Heatmap")
plt.show()
```

#### Practice Task

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic Dataset
titanic = sns.load_dataset("titanic")

# Dark Theme
plt.style.use("dark_background")

# Create Dashboard
fig, axs = plt.subplots(1, 3, figsize=(18, 6), facecolor="#0f172a")
fig.suptitle(
    "Titanic Dataset Analysis",
    fontsize=20,
    color="cyan",
    fontweight="bold"
)

# ====================================================
# 1. Count Plot - Passengers by Class
# ====================================================

sns.countplot(
    data=titanic,
    x="class",
    hue="class",
    palette="viridis",
    legend=False,
    ax=axs[0]
)

axs[0].set_title(
    "Passengers by Class",
    color="white",
    fontsize=14,
    fontweight="bold"
)

axs[0].set_xlabel("Passenger Class")
axs[0].set_ylabel("Count")
axs[0].grid(axis="y", linestyle="--", alpha=0.3)

# ====================================================
# 2. Box Plot - Age by Class and Gender
# ====================================================

sns.boxplot(
    data=titanic,
    x="class",
    y="age",
    hue="sex",
    palette="Set2",
    ax=axs[1]
)

axs[1].set_title(
    "Age by Class and Gender",
    color="white",
    fontsize=14,
    fontweight="bold"
)

axs[1].set_xlabel("Passenger Class")
axs[1].set_ylabel("Age")
axs[1].grid(axis="y", linestyle="--", alpha=0.3)

# ====================================================
# 3. Heatmap - Correlation Matrix
# ====================================================

corr = titanic.select_dtypes(include="number").corr()

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=1,
    linecolor="white",
    ax=axs[2]
)

axs[2].set_title(
    "Correlation Heatmap",
    color="white",
    fontsize=14,
    fontweight="bold"
)

# ====================================================
# Layout
# ====================================================

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
```
