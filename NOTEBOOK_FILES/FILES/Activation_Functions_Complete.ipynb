# Activation Functions in Neural Networks

This notebook explains the most common activation functions:

1. Sigmoid
2. Tanh
3. ReLU
4. Leaky ReLU
5. Linear
6. Softmax

For every function, the notebook contains:
- Definition
- Formula
- Python implementation
- Sample test
- Visualization
- Practical use

## 1. What is an Activation Function?

An **activation function** is a mathematical function applied to the output of a neuron.

It introduces **non-linearity** into a neural network, allowing the network to learn complex relationships and patterns.

Common activation functions include:
- Sigmoid
- Tanh
- ReLU
- Leaky ReLU
- Linear
- Softmax

```python
import numpy as np
import matplotlib.pyplot as plt

# Common input values
x = np.linspace(-6, 6, 500)
```

# 2. Sigmoid

### Definition
Sigmoid converts an input into a value between **0 and 1**.

### Formula
\[
\sigma(x) = \frac{1}{1 + e^{-x}}
\]

### Main use
- Binary classification
- Output probability between 0 and 1

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Sample test
sample = np.array([-2, -1, 0, 1, 2])
print("Input:", sample)
print("Sigmoid:", sigmoid(sample))
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.plot(x, sigmoid(x))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

# 3. Tanh

### Definition
Tanh (Hyperbolic Tangent) converts values into a range from **-1 to +1**.

### Formula
\[
tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}
\]

### Main use
- Hidden layers
- Historically common in recurrent neural networks
- Zero-centered activation

```python
def tanh(x):
    return np.tanh(x)

# Sample test
sample = np.array([-2, -1, 0, 1, 2])
print("Input:", sample)
print("Tanh:", tanh(sample))
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.plot(x, tanh(x))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title("Tanh Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

# 4. ReLU

### Definition
ReLU (Rectified Linear Unit) returns **0 for negative inputs** and the input itself for positive inputs.

### Formula
\[
ReLU(x) = max(0, x)
\]

### Main use
- Very common in hidden layers
- Computationally simple
- Helps neural networks learn non-linear patterns

```python
def relu(x):
    return np.maximum(0, x)

# Sample test
sample = np.array([-2, -1, 0, 1, 2])
print("Input:", sample)
print("ReLU:", relu(sample))
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.plot(x, relu(x))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title("ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

# 5. Leaky ReLU

### Definition
Leaky ReLU is a modified ReLU that allows a small negative output instead of making every negative input zero.

### Formula
\[
f(x) =
\begin{cases}
x, & x > 0 \\
\alpha x, & x \leq 0
\end{cases}
\]

Usually, \(\alpha = 0.01\).

### Main use
- Hidden layers
- Helps reduce the **dying ReLU problem**

```python
def leaky_relu(x, alpha=0.01):
    return np.where(x > 0, x, alpha * x)

# Sample test
sample = np.array([-2, -1, 0, 1, 2])
print("Input:", sample)
print("Leaky ReLU:", leaky_relu(sample))
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.plot(x, leaky_relu(x))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title("Leaky ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

# 6. Linear

### Definition
The Linear activation function returns the input without changing it.

### Formula
\[
f(x) = x
\]

### Main use
- Regression output layers
- Predicting continuous values such as price, temperature, or salary

```python
def linear(x):
    return x

# Sample test
sample = np.array([-2, -1, 0, 1, 2])
print("Input:", sample)
print("Linear:", linear(sample))
```

```python
# Visualization
plt.figure(figsize=(8, 5))
plt.plot(x, linear(x))
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title("Linear Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.show()
```

# 7. Softmax

### Definition
Softmax converts a set of values into probabilities.

All probabilities are between **0 and 1**, and their total is **1**.

### Formula
\[
Softmax(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}
\]

### Main use
- Multi-class classification
- Usually used in the output layer

Example:
- Cat → 0.70
- Dog → 0.20
- Horse → 0.10

Total = 1.00

```python
def softmax(x):
    # Subtracting max improves numerical stability
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

# Sample test
sample = np.array([2.0, 1.0, 0.1])
result = softmax(sample)

print("Input:", sample)
print("Softmax:", result)
print("Sum of probabilities:", np.sum(result))
```

```python
# Visualization
classes = ["Class 1", "Class 2", "Class 3"]
probabilities = softmax(np.array([2.0, 1.0, 0.1]))

plt.figure(figsize=(8, 5))
bars = plt.bar(classes, probabilities)

plt.title("Softmax Output Probabilities")
plt.xlabel("Classes")
plt.ylabel("Probability")

for bar, probability in zip(bars, probabilities):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{probability:.3f}",
        ha="center",
        va="bottom"
    )

plt.ylim(0, 1)
plt.grid(axis="y")
plt.show()
```

# 8. Comparison of Activation Functions

| Function | Range | Main Use |
|---|---|---|
| Sigmoid | 0 to 1 | Binary classification |
| Tanh | -1 to +1 | Hidden layers / RNNs |
| ReLU | 0 to ∞ | Hidden layers |
| Leaky ReLU | -∞ to ∞ | Hidden layers |
| Linear | -∞ to ∞ | Regression |
| Softmax | 0 to 1 | Multi-class classification |

### Easy way to remember

- **Sigmoid** → Binary classification
- **Tanh** → -1 to +1
- **ReLU** → Negative values become 0
- **Leaky ReLU** → Negative values get a small value
- **Linear** → Input stays the same
- **Softmax** → Multiple class probabilities

# 9. Visual Comparison

The following graph compares Sigmoid, Tanh, ReLU, Leaky ReLU, and Linear functions.

```python
plt.figure(figsize=(10, 6))

plt.plot(x, sigmoid(x), label="Sigmoid")
plt.plot(x, tanh(x), label="Tanh")
plt.plot(x, relu(x), label="ReLU")
plt.plot(x, leaky_relu(x), label="Leaky ReLU")
plt.plot(x, linear(x), label="Linear")

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.title("Comparison of Activation Functions")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.legend()
plt.grid(True)
plt.show()
```

# 10. Final Summary

Activation functions are essential because they introduce **non-linearity** into neural networks.

### Key points

**Sigmoid**
- Range: 0 to 1
- Best known for binary classification

**Tanh**
- Range: -1 to +1
- Zero-centered

**ReLU**
- Negative input → 0
- Positive input → same value
- Very common in hidden layers

**Leaky ReLU**
- Keeps a small negative slope
- Helps with dying ReLU

**Linear**
- Output equals input
- Useful for regression

**Softmax**
- Converts multiple scores into probabilities
- Probabilities sum to 1
- Used for multi-class classification
