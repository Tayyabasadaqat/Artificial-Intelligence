#### Data frame in pandas

Data Frames always in 2-Dimensional

```python
data = {
    "Name":['Ali','Ahmed','Sara','Ayesha','Emaan'],
    "Age":[23,24,27,19,32],
    "Maths":[33,45,67,89,90],
    "Computer":[89,78,67,98,67]
}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

#### Data frame using numpy

```python
names = np.random.choice(['Ali','Ahmed','Sara','Ayesha','Emaan'],100)
age=np.random.choice([23,24,27,19,32],100)
maths=np.random.choice([33,45,67,89,90],100)
physics=np.random.choice([45,23,56,78,89],100)
computer=np.random.choice([89,78,67,98,67],100)
english =np.random.choice([56,67,34,19,67],100)
urdu=np.random.choice([67,89,67,54,67],100)

data={
    'Name':names,
    'Age':age,
    'Math':maths,
    'Physics':physics,
    'Computer':computer,
    'English':english,
    'Urdu':urdu
}
df=pd.DataFrame(data)
df
```
