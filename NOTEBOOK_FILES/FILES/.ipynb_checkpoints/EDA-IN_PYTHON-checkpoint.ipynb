```python
# Array(3,4,6)
e=np.array(
    [
        [
            [1,2,3,4,5,6],
            [2,3,4,5,6,7],
            [3,4,5,6,7,8],
            [4,5,6,7,8,9],
        ],
        [
            [9,8,7,6,5,4],
            [8,7,6,5,4,3],
            [7,6,5,4,3,2],
            [6,5,4,3,2,1]
        ],
        [
            [5,6,7,3,4,5],
            [1,2,4,7,9,0],
            [2,5,8,3,4,7],
            [0,3,7,1,9,3]
        ]
    ]
)
e, e.ndim,e.shape
```

```python
e.shape, e.size, e.dtype, e.itemsize, e.nbytes, e.T
```

### Summary
- .ndim = tells dimensions
- .shape = tells shape of the array (3,4,6)
- .size = tells about the number of variablesin array
- .dtype= tells data type of array
- .itemsize = tells how much of a size is of an element
- .nbytes = tells how much space array accquired in memory
- .T =  takes transpose (rows into coloumn and coloumn to rows)

### Stats In Nump

```python
a =np.random.randint(1,20,50)
print(a)
```

```python
# values
print(a.min(),a.max())
```

```python
print(a.sum())
```

```python
# index of min & max values
print(a.argmin(),a.argmax())
```

```python
len(a)
```

```python
print("- "*25)
print(f"{"Statistics":^50}")
print("- "*25)
print(f"{"Min    ":<20}= {np.min(a)}")
print(f"{"Max   ":<20}= {np.max(a)}")
print(f"{"Mean    ":<20}= {np.mean(a)}")
print(f"{"Median    ":<20}= {np.median(a)}")
print(f"{"Variance    ":<20}= {np.var(a):.1F}")
print(f"{"Standard Deviation  ":<20}= {np.std(a):.3F}")
print(f"{"Quartiles  ":<20}= {np.percentile(a,[0,25,50,75,100])}")

print("- "*25)
```
