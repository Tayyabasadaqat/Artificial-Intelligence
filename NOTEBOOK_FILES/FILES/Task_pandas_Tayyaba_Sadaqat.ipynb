## Solution

* Drop all rows that contain at least one null value and report how many rows remain.
* Fill missing values in the age column with the mean age, and missing values in salary with the median salary.
* Fill missing values in the department column with the mode (most frequent department).
* Group the data by department and calculate the average salary for each group (ignoring nulls).
* Find all rows where age is greater than 40 and salary is above 80,000.
* Sort the DataFrame by salary in descending order and display the top 10 records.
* Create a new column salary_category that labels each row as "Low" (<50,000), "Medium" (50,000–90,000), or "High" (>90,000) based on the salary column, handling nulls appropriately.

### Drop all rows that contain at least one null value and report how many rows remain.

```python
df_no_null = df.dropna()

print("Rows remaining:", len(df_no_null))
df_no_null.head()
```

### Fill missing values in the age column with the mean age, and missing values in salary with the median salary.

```python
df['age'] = df['age'].fillna(df['age'].mean())
df['salary'] = df['salary'].fillna(df['salary'].median())
df.head()
```

### Fill missing values in the department column with the mode (most frequent department).

```python
df['department'] = df['department'].fillna(df['department'].mode()[0])
df.head()
```

### Group the data by department and calculate the average salary for each group (ignoring nulls).

```python
avg_salary = df.groupby('department')['salary'].mean()
print(avg_salary)
```

### Find all rows where age is greater than 40 and salary is above 80,000.

```python
result = df[(df['age'] > 40) & (df['salary'] > 80000)]
print(result)
```

### Sort the DataFrame by salary in descending order and display the top 10 records.

```python
top10 = df.sort_values(by='salary', ascending=False)
top10.head(10)
```

### Create a new column salary_category that labels each row as "Low" (<50,000), "Medium" (50,000–90,000), or "High" (>90,000) based on the salary column, handling nulls appropriately.

```python
def salary_category(salary):
    if pd.isna(salary):
        return "Missing"
    elif salary < 50000:
        return "Low"
    elif salary <= 90000:
        return "Medium"
    else:
        return "High"
        
df['salary_category'] = df['salary'].apply(salary_category)
df[['salary', 'salary_category']].head()
```
