<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 1: Define a Dateset
</h2>

In this task, we import the NumPy library and generate a dataset containing 1000 random integers between 1 and 10,000.

```python
import numpy as np
np.random.seed(13)     # Makes the random numbers reproducible
data = np.random.randint(1, 10001, 1000)

print("Dataset Shape:", data.shape)
print(data)
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 2: Find the 10 Largest Values
</h2>

This task finds the 10 largest numbers in the dataset.
NumPy provides the `sort()` function to arrange values in ascending order.
After sorting, we select the last 10 values and reverse them to display them from largest to smallest.

```python
largest_10 = np.sort(data)

print("10 Largest Values:")
print(largest_10[-10:])
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 3: Find the 10 Smallest Values
</h2>

In this task, we sort the dataset in ascending order and display the first 10 values. Since the data is sorted from smallest to largest, the first 10 values are the smallest numbers in the dataset.

```python
sorted_data = np.sort(data)

print("10 Smallest Values:")
print(sorted_data[:10])
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 4: Find All Unique Values
</h2>

A unique value is a value that appears in the dataset.
NumPy provides the `unique()` function which returns all distinct values without repetition.

```python
unique_values = np.unique(data)

print("Unique Values:")
print(unique_values)
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 5: Calculate the Frequency of Each Unique Value
</h2>

This task counts how many times each unique value appears in the dataset.

```python
unique_values, frequency = np.unique(data, return_counts=True)

for i in range(len(unique_values)):
    print(unique_values[i], "appears", frequency[i], "times")
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 6: Count Even and Odd Numbers
</h2>

In this task, we count how many even and odd numbers are present in the dataset.
- Even numbers are divisible by 2.
- Odd numbers are not divisible by 2.

```python
even_count = np.sum(data % 2 == 0)
odd_count = np.sum(data % 2 != 0)

print("Number of Even Values:", even_count)
print("Number of Odd Values:", odd_count)
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 7: Find Numbers Divisible by 7
</h2>

This task finds all the numbers in the dataset that are divisible by 7.
A number is divisible by 7 if the remainder after dividing by 7 is equal to 0.

```python
divisible_by_7 = data[data % 7 == 0]

print("Numbers divisible by 7:")
print(divisible_by_7)
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 8: Percentage of Values Greater Than 5000
</h2>

This task calculates the percentage of numbers in the dataset that are greater than 5000.

```python
count = np.sum(data > 5000)

percentage = (count / len(data)) * 100

print("Percentage of values greater than 5000:", percentage, "%")
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 9: Display Dataset in Ascending Order
</h2>

This task sorts the dataset from the smallest value to the largest value using the `np.sort()` function.

```python
ascending = np.sort(data)

print("Dataset in Ascending Order:")
print(ascending)
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 10: Display Dataset in Descending Order
</h2>

This task sorts the dataset from the largest value to the smallest value by reversing the ascending sorted array.

```python
descending = np.sort(data)[::-1]

print("Dataset in Descending Order:")
print(descending)
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 11: Find the Index of the Maximum Value
</h2>

This task finds the position (index) of the largest value in the dataset.

```python
max_index = np.argmax(data)

print("Index of Maximum Value:", max_index)
print("Maximum Value:", data[max_index])
```

<h2 style="background-color:#1565C0; color:white; padding:10px; border-radius:8px;">
Task 12: Find the Index of the Minimum Value
</h2>

This task finds the position (index) of the smallest value in the dataset.

```python
min_index = np.argmin(data)

print("Index of Minimum Value:", min_index)
print("Minimum Value:", data[min_index])
```
