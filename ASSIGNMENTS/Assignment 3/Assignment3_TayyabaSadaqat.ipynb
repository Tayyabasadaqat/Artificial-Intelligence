## Section A: Short Theory Questions

### Q1. Differentiate between a class and an object with a real-world example.

A **class** is a blueprint or template that defines the properties (attributes) and behaviors (methods) of a particular type of object. An **object** is a real instance created from that class.

### Real-world Example: Car
- **Class:** Car
- **Objects:** Toyota Corolla, Honda Civic, Suzuki Alto

The class defines attributes such as color, model, and engine capacity, while each object has its own specific values for those attributes. Multiple objects can be created from the same class, but each object maintains its own data.

### Q2. What is the difference between instance variables and class variables?

**Instance variables** belong to a specific object, while **class variables** belong to the class itself and are shared by all objects.

### Instance Variables
- Created inside the constructor using `self`
- Each object has its own copy
- Changes affect only that object

### Class Variables
- Defined directly inside the class
- Shared among all objects
- Changes are reflected across all instances unless overridden

Example: In a `Perfume` class, `name` and `price` are instance variables, while `total_perfumes` is a class variable that counts all perfume objects created.

### Q3. Explain the difference between Single, Multilevel, Multiple, and Hierarchical inheritance.

Inheritance allows one class to acquire properties and methods from another class.

### Single Inheritance
One child class inherits from one parent class.

### Multilevel Inheritance
A child inherits from a parent, and another class inherits from that child, forming a chain.

### Multiple Inheritance
A child class inherits from more than one parent class.

### Hierarchical Inheritance
Multiple child classes inherit from the same parent class.

These inheritance types help promote **code reuse**, **maintainability**, and **extensibility** in object-oriented programs.

### Q4. What is encapsulation, and how is it implemented in Python using access specifiers?

**Encapsulation** is the process of hiding internal data and restricting direct access to it. It protects an object's state and allows controlled access through methods.

Python uses naming conventions for access control:

- **Public (`name`)** → accessible from anywhere.
- **Protected (`_name`)** → intended for internal use within the class and subclasses.
- **Private (`__name`)** → Python performs **name mangling** to make direct access difficult.

Example:
```python
class Perfume:
    def __init__(self):
        self.name = "Rose"
        self._supplier = "Aroma Ltd"
        self.__stock = 50
```

The private attribute can still be accessed using its mangled name, such as `_Perfume__stock`.

### Q5. What is the difference between method overriding and operator overloading?

### Method Overriding
- Occurs when a child class provides its own implementation of a method already defined in the parent class.
- Used to achieve **runtime polymorphism**.

### Operator Overloading
- Occurs when built-in operators such as `+`, `-`, `*`, or `==` are given custom behavior for user-defined objects.
- Implemented using special methods like `__add__()`, `__sub__()`, and `__eq__()`.

Example:
- **Method Overriding:** `GiftSet.calculate_final_price()` overrides the abstract method from `Product`.
- **Operator Overloading:** `perfume1 + perfume2` creates a `ComboPack` object with a combined price.

## Task 1: Create a Perfume Class

### Logic and Approach
The first step is to create a basic `Perfume` class that stores information about a perfume product. ## Task 2: Add a Class Variable

### Logic and Approach
A **class variable** is shared by all objects of a class. Here, `total_perfumes` keeps track of how many perfume objects have been created. Every time the constructor runs, the counter is automatically incremented.

This approach is useful for inventory statistics, analytics, and monitoring how many products are added to the system without manually updating a counter.


````python
class Perfume:
    total_perfumes = 0  # Class variable

    def __init__(self, name, brand, price, volume_ml):
        self.name = name
        self.brand = brand
        self.price = price
        self.volume_ml = volume_ml

        # Increment counter whenever an object is created
        Perfume.total_perfumes += 1

    def display_details(self):
        print(f"{self.name} | {self.brand} | Rs.{self.price} | {self.volume_ml}ml")


# Creating objects
p1 = Perfume("Rose Mist", "Aroma House", 4500, 50)
p2 = Perfume("Oud Royal", "Luxury Scents", 8500, 100)
p3 = Perfume("Citrus Bloom", "Fresh Aura", 5200, 75)

print("Total perfumes created:", Perfume.total_perfumes)
````

## Task 3: Inheritance with Product, Perfume, and Accessory

### Logic and Approach
A `Product` base class is created to store attributes common to all shop items, such as **name** and **price**. The `Perfume` and `Accessory` classes inherit from `Product`, allowing them to reuse existing code while adding their own specialized attributes.

This demonstrates **single inheritance** and shows how inheritance reduces code duplication by placing shared functionality in a parent class.


````python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class Perfume(Product):
    def __init__(self, name, price, brand, volume_ml):
        super().__init__(name, price)
        self.brand = brand
        self.volume_ml = volume_ml


class Accessory(Product):
    def __init__(self, name, price, accessory_type):
        super().__init__(name, price)
        self.accessory_type = accessory_type


# Objects
perfume = Perfume("Rose Mist", 4500, "Aroma House", 50)
accessory = Accessory("Luxury Gift Box", 1200, "Gift Packaging")

print(perfume.name, perfume.price, perfume.brand)
print(accessory.name, accessory.price, accessory.accessory_type)
````

## Task 4: Multiple Inheritance and MRO

### Logic and Approach
A `Discountable` class contains discount-related behavior, while `Perfume` contains perfume-related data. The `PremiumPerfume` class inherits from **both** classes, demonstrating **multiple inheritance**.

The **Method Resolution Order (MRO)** shows the order in which Python searches parent classes when a method is called. Understanding MRO is important because it prevents ambiguity when multiple parent classes contain methods with the same name.


````python
class Discountable:
    def apply_discount(self, percentage):
        return self.price - (self.price * percentage / 100)


class Perfume:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class PremiumPerfume(Discountable, Perfume):
    def __init__(self, name, price):
        Perfume.__init__(self, name, price)


premium = PremiumPerfume("Royal Oud", 12000)

print("Original Price:", premium.price)
print("After 15% Discount:", premium.apply_discount(15))

# MRO
print("\nMethod Resolution Order:")
for cls in PremiumPerfume.__mro__:
    print(cls.__name__)
````

## Task 5: Encapsulation and Name Mangling

### Logic and Approach
This task demonstrates three access levels:

- **Public:** `name`
- **Protected:** `_supplier_name`
- **Private:** `__stock_quantity`

Python performs **name mangling** for private attributes by internally renaming them. Direct access using `object.__stock_quantity` will fail, but the attribute can be accessed using its mangled name `_ClassName__attribute`.

This illustrates how encapsulation helps protect sensitive data such as inventory stock from accidental modification.


````python
class Perfume:
    def __init__(self, name, supplier, stock):
        self.name = name                    # Public
        self._supplier_name = supplier     # Protected
        self.__stock_quantity = stock      # Private


p = Perfume("Rose Mist", "Aroma Suppliers", 40)

# Public access
print("Public:", p.name)

# Protected access (possible, but not recommended)
print("Protected:", p._supplier_name)

# Name mangling demonstration
print("Private via name mangling:", p._Perfume__stock_quantity)
````

## Task 6: Operator Overloading with __add__()

### Logic and Approach
The `+` operator is overloaded using the `__add__()` method. When two `Perfume` objects are added together, a new `ComboPack` object is created containing a combined description and total price.

This provides a natural and readable syntax for creating perfume bundles, which is a common feature in retail systems.


````python
class ComboPack:
    def __init__(self, description, total_price):
        self.description = description
        self.total_price = total_price

    def show(self):
        print(f"Combo: {self.description}")
        print(f"Total Price: Rs.{self.total_price}")


class Perfume:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __add__(self, other):
        description = f"{self.name} + {other.name}"
        total_price = self.price + other.price
        return ComboPack(description, total_price)


p1 = Perfume("Rose Mist", 4500)
p2 = Perfume("Oud Royal", 8500)

combo = p1 + p2
combo.show()
````

## Task 7: Abstract Class and calculate_final_price()

### Logic and Approach
An **abstract class** defines a common interface that all product types must follow. The abstract method `calculate_final_price()` forces every subclass to provide its own implementation.

- `Perfume` applies **10% tax**.
- `GiftSet` applies **5% discount**.

This ensures consistency across product types while allowing different pricing rules for different categories.


````python
from abc import ABC, abstractmethod


class Product(ABC):
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @abstractmethod
    def calculate_final_price(self):
        pass


class Perfume(Product):
    def calculate_final_price(self):
        # Add 10% tax
        return self.price * 1.10


class GiftSet(Product):
    def calculate_final_price(self):
        # Apply 5% discount
        return self.price * 0.95


perfume = Perfume("Rose Mist", 5000)
gift = GiftSet("Luxury Gift Set", 8000)

print("Perfume Final Price:", perfume.calculate_final_price())
print("Gift Set Final Price:", gift.calculate_final_price())
````

## Task 8: Dynamic Polymorphism

### Logic and Approach
Dynamic polymorphism occurs when the same method call produces different behavior depending on the object's actual type at runtime.

A `Product`-type reference variable is used to point first to a `Perfume` object and then to a `GiftSet` object. Calling `calculate_final_price()` executes the appropriate subclass implementation automatically.

This demonstrates **runtime polymorphism**, one of the most important features of object-oriented programming.


````python
from abc import ABC, abstractmethod


class Product(ABC):
    def __init__(self, name, price):
        self.name = name
        self.price = price

    @abstractmethod
    def calculate_final_price(self):
        pass


class Perfume(Product):
    def calculate_final_price(self):
        return self.price * 1.10


class GiftSet(Product):
    def calculate_final_price(self):
        return self.price * 0.90


# Product-type reference variable
products = [
    Perfume("Rose Mist", 5000),
    GiftSet("Luxury Gift Set", 8000)
]

for product in products:
    print(f"{product.name} -> Final Price: Rs.{product.calculate_final_price():.2f}")
````
