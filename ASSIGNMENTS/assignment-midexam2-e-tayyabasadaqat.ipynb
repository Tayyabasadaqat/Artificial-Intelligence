# Car Price Prediction using Machine Learning

## Project Objective

The objective of this project is to build a machine learning regression pipeline to predict the selling price of cars in Pakistani Rupees (PKR).

The dataset contains information about car brand, model, year, mileage, fuel type, transmission, engine capacity, city, color, number of previous owners, registration status, and selling price.

The project includes data exploration, data cleaning, feature engineering, preprocessing, model training, evaluation, and visualization.

## 9. Handling Missing Values

Missing numerical values are handled using the median because the median is less sensitive to extreme values than the mean.

Missing categorical values are handled using the most frequent value (mode).

Rows with a missing target value are removed because the target price must be known in order to train a supervised regression model.

```python
# Remove rows where the target value is missing
df = df.dropna(subset=["Price_PKR"])

# Numerical columns
numeric_cols = [
    "Year",
    "Mileage_km",
    "Engine_CC",
    "Num_Owners"
]

# Categorical columns
categorical_cols = [
    "Brand",
    "Model",
    "Fuel_Type",
    "Transmission",
    "City",
    "Color",
    "Registered"
]

# Fill numerical missing values with median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values with mode
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

## 13. Feature Engineering — Car Age

The manufacturing year is converted into car age because the age of a vehicle can have a direct relationship with its selling price.

A newer car generally tends to have a higher market value than an older car, although other factors also affect price.

```python
df["Car_Age"] = 2026 - df["Year"]

df[["Year", "Car_Age"]].head()
```

## Removing the Original Year Feature

The original Year feature is removed after creating Car_Age because Car_Age provides a more directly interpretable representation of vehicle age.

```python
df = df.drop("Year", axis=1)
```

## 14. Separating Features and Target

The target variable is Price_PKR because the objective of the project is to predict the selling price of a car.

All other relevant columns are used as input features.

```python
X = df.drop("Price_PKR", axis=1)
y = df["Price_PKR"]

print("Features:")
print(X.columns.tolist())

print("\nTarget:")
print("Price_PKR")
```

## 15. Identifying Numerical and Categorical Features

Numerical and categorical features are identified separately because they require different preprocessing techniques before being provided to machine learning models.

```python
numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
```

## 16. Train-Test Split

The dataset is divided into training and testing sets.

The training set is used to train the machine learning models, while the testing set is kept unseen during training and is used to evaluate how well the models generalize to new data.

An 80:20 split is used, with 80% of the data for training and 20% for testing.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

## 17. Feature Preprocessing

The numerical features are standardized using StandardScaler so that numerical variables are placed on comparable scales.

Categorical features are converted into numerical representations using One-Hot Encoding.

A ColumnTransformer is used to apply the appropriate preprocessing to each feature type.

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

## 18. Linear Regression

Linear Regression is used as a baseline regression model. It attempts to model the relationship between the input features and the car price using a linear relationship.

```python
linear_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

linear_model.fit(X_train, y_train)
```

## Linear Regression Predictions

The trained Linear Regression model is used to predict prices for the unseen testing data.

```python
linear_pred = linear_model.predict(X_test)

print(linear_pred[:10])
```

## 19. Ridge Regression

Ridge Regression is a regularized version of Linear Regression. It adds a penalty to large model coefficients and can help reduce overfitting, especially when features are correlated.

```python
ridge_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Ridge(alpha=1.0))
])

ridge_model.fit(X_train, y_train)

ridge_pred = ridge_model.predict(X_test)
```

## 20. Random Forest Regression

Random Forest Regressor is an ensemble learning algorithm that combines multiple decision trees to produce predictions.

It can capture non-linear relationships between car characteristics and price and is therefore suitable for this regression problem.

```python
rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])

rf_model.fit(X_train, y_train)

rf_pred = rf_model.predict(X_test)
```

## 21. Model Evaluation

The regression models are evaluated using Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² score.

For MAE, MSE, and RMSE, lower values indicate smaller prediction errors.

For R², a higher value generally indicates better explanatory performance.

```python
def evaluate_model(model_name, y_true, predictions):
    mae = mean_absolute_error(y_true, predictions)
    mse = mean_squared_error(y_true, predictions)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, predictions)

    return {
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2
    }


results = []

results.append(
    evaluate_model("Linear Regression", y_test, linear_pred)
)

results.append(
    evaluate_model("Ridge Regression", y_test, ridge_pred)
)

results.append(
    evaluate_model("Random Forest", y_test, rf_pred)
)

results_df = pd.DataFrame(results)

results_df
```

## 22. Model Comparison

The models are compared using their evaluation metrics. The preferred regression model should generally have lower MAE, MSE, and RMSE and a higher R² score.

```python
results_df.sort_values(by="R2", ascending=False)
```

## 23. Selecting the Best Model

The model with the highest R² score is selected as the best-performing model for this dataset.

```python
best_model_name = results_df.loc[
    results_df["R2"].idxmax(),
    "Model"
]

print("Best Model:", best_model_name)
```
