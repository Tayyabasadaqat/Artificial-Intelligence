## Descriptive Statistics

### 1. Quantitative vs Qualitative

- **Quantitative data** are numerical values (e.g., marks, height). 
- **Qualitative data** are categories (e.g., gender, color).

### 2. Mean Median Mode

- Mean is the arithmetic average. 
- Median is the middle value after sorting. Median is preferred when data contains outliers.
- Mode is the most frequent value. 

### 3. Variance & Standard Deviation

- Variance measures spread around the mean. 
- Standard deviation is the square root of variance and is in the original units.

### 4. Z-score

A Z-score shows how many standard deviations a value is from the mean. Values beyond about ±3 are often considered outliers.

## Practical 1
Calculate mean, median and mode manually and using built-in functions.

```python
import statistics as st
marks=[78,85,92,67,74,88,90,81,76,85]
print("Marks:",marks)
manual_mean=sum(marks)/len(marks)
s=sorted(marks)
manual_median=(s[4]+s[5])/2
from collections import Counter
manual_mode=Counter(marks).most_common(1)[0][0]
print(manual_mean,manual_median,manual_mode)
print(st.mean(marks),st.median(marks),st.mode(marks))
```

## Practical 2
Variance, standard deviation and coefficient of variation.

```python
import statistics as st
mean=st.mean(marks)
var=st.variance(marks)
std=st.stdev(marks)
cv=(std/mean)*100
print(var,std,cv)
```

## Practical 3
Z-scores and outlier detection.

```python
zs=[(x-mean)/std for x in marks]
for m,z in zip(marks,zs):
    print(m,round(z,2),"Outlier" if abs(z)>3 else "")
```

## Practical 4
Correlation coefficient.

```python
import numpy as np
study_hours=np.array([2,3,4,5,6,7,8,4,5,6])
marks_np=np.array(marks)
print(np.corrcoef(study_hours,marks_np)[0,1])
```
